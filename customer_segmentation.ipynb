```python
import pandas as pd

df = pd.read_csv("Mall_Customers.csv")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)"
)

plt.title("Customer Data")
plt.show()
```

```python
from sklearn.cluster import KMeans

features = df[[
    "Annual Income (k$)",
    "Spending Score (1-100)"
]]

model = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df["Segment"] = model.fit_predict(features)

df.head()
```

```python
sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Segment",
    palette="Set1",
    s=100
)

plt.title("Customer Segmentation")
plt.show()
```

```python
df.groupby("Segment")[[
    "Age",
    "Annual Income (k$)",
    "Spending Score (1-100)"
]].mean().round(2)
```
